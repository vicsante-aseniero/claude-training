```python
# Load env variables
from dotenv import load_dotenv
import os

# Reads .env file in the project root and injects into os.environ
load_dotenv()

# Optional: confirm the key was found and with exposing it
print("API key loaded:", bool(os.environ.get("ANTHROPIC_API_KEY")))
```

```python
# Create an API client
import anthropic
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-5"
```

```python
# Check or verify if we can use Claude API with our current API key

from anthropic import Anthropic, AuthenticationError, BadRequestError, RateLimitError, APIConnectionError

try:
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=[{"role": "user", "content": "Hello"}]
    )
    print(message.content[0].text)

except anthropic.AuthenticationError:
    # API key is missing or invalid
    print("Check your ANTHROPIC_API_KEY.")

except anthropic.BadRequestError as e:
    # Could be billing, malformed request, or invalid parameters
    # The error body tells you which — inspect e.body or str(e)
    print(f"Bad request: {e}")

except anthropic.RateLimitError:
    # Too many requests — back off and retry
    print("Rate limited. Slow down your request rate.")

except anthropic.APIConnectionError:
    # Network issue — couldn't reach the API at all
    print("Connection failed. Check your network.")
```

```python
# Make a request

def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages,
    )
    return message.content[0].text

# Make a starting list of messages
messages = []

# Add in the initial user questions of "Define quantum computing in one sentence"
add_user_message(messages, "Define quantum computing in one sentence")

# Pass the list of messages into 'chat' to get an answer
answer = chat(messages)

# Take the answer and add it as an assistant message into our list
add_assistant_message(messages, answer)

# Add in the user's follow-up questions
add_user_message(messages, "Write another sentence")

# Call chat again with the list of messages to get a final answer
answer = chat(messages)

answer
```
