```python
# Load env variables and create client
from dotenv import load_dotenv
import os

# Reads .env file in the project root and injects into os.environ
load_dotenv()

# Optional: confirm the key was found and with exposing it
print("API key loaded:", bool(os.environ.get("ANTHROPIC_API_KEY")))

# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-5"
```

```python
# Helper functions
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None, temperature=1.0):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "temperature": temperature,
    }

    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text
```

```python
messages = []

add_user_message(
    messages,
    "Write a 1 sentence description of a fake database"
)

stream = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=messages,
    stream=True
)

for event in stream:
    print(event)
```
